```python
import pandas as pd
```

```python
df = pd.read_csv("student_data.csv")
```

```python
# 1. Department with highest average score
department_average = df.groupby("Department")["Final_Marks"].mean()

print("Department-wise Average Marks:")
print(department_average)

print("\nHighest Average:")
print(department_average.idxmax(),
      department_average.max())
```

```python
# 2. Does attendance affect final marks?
attendance_correlation = df["Attendance"].corr(df["Final_Marks"])

print("\nAttendance vs Final Marks Correlation:")
print(attendance_correlation)
```

```python
# 3. Relationship between study hours and marks
study_correlation = df["Study_Hours"].corr(df["Final_Marks"])

print("\nStudy Hours vs Final Marks Correlation:")
print(study_correlation)
```

```python
# 4. Department with highest failure rate
failure_rate = (
    df.groupby("Department")["Result"]
      .apply(lambda x: (x == "Fail").mean() * 100)
)

print("\nDepartment-wise Failure Rate:")
print(failure_rate)

print("\nHighest Failure Rate:")
print(failure_rate.idxmax(),
      failure_rate.max())
```

```python
# 5. Students with attendance below 75%
below_75 = df[df["Attendance"] < 75]

print("\nStudents with Attendance Below 75%:")
print(len(below_75))
```

```python
# 6. Percentage of students who passed
pass_percentage = (df["Result"] == "Pass").mean() * 100

print("\nPass Percentage:")
print(pass_percentage)
```

```python

# 7. Factors associated with final marks
numeric_columns = [
    "Attendance",
    "Study_Hours",
    "Assignment_Score",
    "Midterm_Score",
    "Final_Marks"
]

correlation = df[numeric_columns].corr()

print("\nCorrelation with Final Marks:")
print(correlation["Final_Marks"].sort_values(ascending=False))
```
