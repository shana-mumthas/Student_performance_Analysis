```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = pd.read_csv("student_performance_500.csv")
```

```python
sns.histplot(data=data,x="Final_Mark",kde=True)

plt.title("Final Mark Distribution")
plt.xlabel("Final Mark")
plt.ylabel("Number of Students")

plt.show()
```

```python
sns.boxplot(
    data=data,
    x="Department",
    y="Final_Mark"
)

plt.title("Final Marks by Department")
plt.xlabel("Department")
plt.ylabel("Final Mark")

plt.show()
```

```python
sns.countplot(
    data=data,
    x="Department"
)

plt.title("Number of Students by Department")
plt.xlabel("Department")
plt.ylabel("Number of Students")

plt.show()
```

```python
plt.figure(figsize=(6, 5))

sns.countplot(
    data=data,
    x="Result"
)

plt.title("Pass vs Fail")
plt.xlabel("Result")
plt.ylabel("Number of Students")

plt.show()
```

```python
sns.scatterplot(
    data=data,
    x="Study_Hours",
    y="Final_Mark"
)

plt.title("Study Hours vs Final Marks")
plt.xlabel("Study Hours")
plt.ylabel("Final Mark")

plt.show()
```

```python
numeric_data = data[
    [
        "Age",
        "Attendance",
        "Assignment_Score",
        "Internal_Mark",
        "Study_Hours",
        "Backlogs",
        "Final_Mark"
    ]
]
correlation = numeric_data.corr()

print(correlation)

sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f"
)

plt.title("Correlation Heatmap")

plt.show()
```
