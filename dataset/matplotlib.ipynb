 Matplotlib
Create charts such as:

Student marks distribution

Department-wise average marks

Attendance distribution

Pass vs Fail

Study hours vs final marks


```python
import pandas as pd
import matplotlib.pyplot as plt

data =pd.read_csv("student_performance_500.csv")
```

```python
plt.figure(figsize=(8, 5))

plt.hist(
    data["Final_Mark"],
    bins=10,
    edgecolor="black"
)

plt.title("Student Final Marks Distribution")
plt.xlabel("Final Mark")
plt.ylabel("Number of Students")

plt.show()
```

```python
department_average = data.groupby(
    "Department"
)["Final_Mark"].mean()

print(department_average)
```

```python
plt.figure(figsize=(8, 5))

plt.bar(
    department_average.index,
    department_average.values
)

plt.title("Department-wise Average Final Marks")
plt.xlabel("Department")
plt.ylabel("Average Final Mark")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.hist(
    data["Attendance"],
    bins=10,
    edgecolor="black"
)

plt.title("Student Attendance Distribution")
plt.xlabel("Attendance (%)")
plt.ylabel("Number of Students")

plt.show()
```

```python
result_count = data["Result"].value_counts()

print(result_count)

plt.bar(
    result_count.index,
    result_count.values
)

plt.title("Pass vs Fail")
plt.xlabel("Result")
plt.ylabel("Number of Students")

plt.show()
```

```python
plt.scatter(
    data["Study_Hours"],
    data["Final_Mark"]
)

plt.title("Study Hours vs Final Marks")
plt.xlabel("Study Hours")
plt.ylabel("Final Mark")

plt.show()
```
