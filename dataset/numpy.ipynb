```python
import numpy as np
import pandas as pd
```

```python
data=pd.read_csv('student_performance_500.csv')
data
```

```python
marks=data['Final_Mark'].to_numpy()
print(marks)
```

```python
attendance=data['Attendance'].to_numpy()
study_hours=data['Study_Hours'].to_numpy()
print(study_hours)
print(attendance)
```

```python
mark_mean=np.mean(marks)
print(mark_mean)
```

```python
mark_median=np.median(marks)
print('median:',mark_median)
```

```python
std_marks=np.std(marks)
print("Standard Deviation:", std_marks)
```

```python
minimum = np.min(marks)

print("Minimum Mark:", minimum)
```

```python
maximum = np.max(marks)

print("Maximum Mark:", maximum)
```

```python
p50 = np.percentile(marks, 50)

print("50th Percentile:", p50)
```

```python
attendance = data["Attendance"].to_numpy()
final_marks = data["Final_Mark"].to_numpy()

correlation = np.corrcoef(
    attendance,
    final_marks
)[0, 1]

print("Attendance vs Final Mark:", correlation)
```
