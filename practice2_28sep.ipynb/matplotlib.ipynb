```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data=pd.read_csv('student_performance_matplotlib.csv')
plt.plot(data["Math"])


plt.title('math marks of all students')
plt.xlabel('students')
plt.ylabel(' mathsmark')
plt.show()
```

```python
# Create a line plot comparing Math and Science marks of all students. 
plt.plot(data["Math"],marker="s",label="maths mark")
plt.plot(data["Science"],marker="o",label="science mark")

plt.title("math vs science marks")
plt.xlabel("students")
plt.ylabel("mark")

plt.legend()
plt.show()



```

```python
#3. Create a line plot comparing Math, Science, and English marks of all students.
plt.plot(data["Math"], marker="o", label="Math")
plt.plot(data["Science"], marker="o", label="Science")
plt.plot(data["English"], marker="o", label="English")

plt.title("Math, Science and English Marks")
plt.xlabel("Student Number")
plt.ylabel("Marks")

plt.legend()
plt.show()
```

```python
#4. Create a bar chart showing the Math marks of each student. 

plt.bar(data.index,data["Math"])
plt.title('Math marks of each student')
plt.xlabel("student")
plt.ylabel("maths")
plt.show()
```

```python
#4. Create a bar chart showing the Science marks of each student. 
plt.bar(data.index,data["Science"])

plt.title("Science Marks of Each Student")
plt.xlabel("Student")
plt.ylabel("Science Marks")

plt.show()
```

```python
#4. Create a horizontal bar chart showing the English marks of each student. 
plt.barh(data.index,data["English"])

plt.title("English Marks of Each Student")
plt.xlabel("English Marks")
plt.ylabel("Student")

plt.show()
```

```python
#4. Create a grouped bar chart comparing Math and Science marks for each student. 
plt.bar(data.index - 0.2, data["Math"], width=0.4, label="Math")
plt.bar(data.index + 0.2, data["Science"], width=0.4, label="Science")

plt.title("Math vs Science Marks")
plt.xlabel("Student")
plt.ylabel("Marks")

plt.legend()
plt.show()
```

```python
gender_count = data["Gender"].value_counts()

plt.pie(gender_count, labels=gender_count.index, autopct="%1.1f%%")

plt.title("Gender Distribution")
plt.show()
```

```python
data["Study_Category"] = pd.cut(
    data["Study_Hours"],
    bins=[0, 2, 4, 6, 10],
    labels=["0-2 Hours", "2-4 Hours", "4-6 Hours", "6-10 Hours"]
)

study_count = data["Study_Category"].value_counts().sort_index()

plt.pie(study_count, labels=study_count.index, autopct="%1.1f%%")

plt.title("Students by Study-Hour Category")
plt.show()
```

```python
average_marks = [
    data["Math"].mean(),
    data["Science"].mean(),
    data["English"].mean()
]

subjects = ["Math", "Science", "English"]

plt.pie(average_marks, labels=subjects, autopct="%1.1f%%")

plt.title("Average Marks of Subjects")
plt.show()
```

```python
fig, ax = plt.subplots(2, 2, figsize=(12, 8))

# Math - Line plot
ax[0, 0].plot(data["Student"], data["Math"], marker="o")
ax[0, 0].set_title("Math Marks")
ax[0, 0].set_xlabel("Student")
ax[0, 0].set_ylabel("Marks")

# Science - Line plot
ax[0, 1].plot(data["Student"], data["Science"], marker="o")
ax[0, 1].set_title("Science Marks")
ax[0, 1].set_xlabel("Student")
ax[0, 1].set_ylabel("Marks")

# English - Bar chart
ax[1, 0].bar(data["Student"], data["English"])
ax[1, 0].set_title("English Marks")
ax[1, 0].set_xlabel("Student")
ax[1, 0].set_ylabel("Marks")

# Gender - Pie chart
gender_count = data["Gender"].value_counts()
ax[1, 1].pie(
    gender_count,
    labels=gender_count.index,
    autopct="%1.1f%%"
)
ax[1, 1].set_title("Gender Distribution")

plt.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots(1, 2, figsize=(12, 5))

# Math vs Science - Line plot
ax[0].plot(data["Student"], data["Math"], marker="o", label="Math")
ax[0].plot(data["Student"], data["Science"], marker="s", label="Science")

ax[0].set_title("Math vs Science")
ax[0].set_xlabel("Student")
ax[0].set_ylabel("Marks")
ax[0].legend()

# Math - Bar chart
ax[1].bar(data["Student"], data["Math"])

ax[1].set_title("Math Marks")
ax[1].set_xlabel("Student")
ax[1].set_ylabel("Marks")

plt.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots(2, 2, figsize=(12, 8))

# Math
ax[0, 0].plot(data["Student"], data["Math"], marker="o")
ax[0, 0].set_title("Math Marks")
ax[0, 0].set_xlabel("Student")
ax[0, 0].set_ylabel("Marks")

# Science
ax[0, 1].plot(data["Student"], data["Science"], marker="o")
ax[0, 1].set_title("Science Marks")
ax[0, 1].set_xlabel("Student")
ax[0, 1].set_ylabel("Marks")

# English
ax[1, 0].bar(data["Student"], data["English"])
ax[1, 0].set_title("English Marks")
ax[1, 0].set_xlabel("Student")
ax[1, 0].set_ylabel("Marks")

# Attendance
ax[1, 1].plot(data["Student"], data["Attendance"], marker="o")
ax[1, 1].set_title("Attendance")
ax[1, 1].set_xlabel("Student")
ax[1, 1].set_ylabel("Attendance")

plt.tight_layout()
plt.show()
```
